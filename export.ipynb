```python
from ultralytics import YOLO

model = YOLO(r"C:\AI\sampah_sungai\Hasil\yolo11n-seg(v1)\weights\best.pt")  # load a custom trained model

# Export the model
model.export(
    format="onnx",         # Export to ONNX
    imgsz=640,             # Ukuran input, bisa disesuaikan
    dynamic=True,          # Dynamic shape ONNX (bagus untuk CPU inference berbagai ukuran)
    simplify=True,         # Sederhanakan ONNX graph
    half=False,            # Jangan pakai FP16 di CPU (FP16 biasanya tidak didukung)
    int8=False,            # INT8 hanya relevan jika kamu punya kalibrasi dan edge-device
    device="cpu"           # Pastikan menggunakan CPU saat export (meski ini opsional saat export)
)
```
